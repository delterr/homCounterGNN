```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
tree1 = list(nx.generators.nonisomorphic_trees(7))[-1]
tree2 = list(nx.generators.nonisomorphic_trees(8))[-1]
tree3 = list(nx.generators.nonisomorphic_trees(9))[-1]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
 
trees = [tree1, tree2, tree3]
titles = ['7-node tree', '8-node tree', '9-node tree']
 
for ax, tree, title in zip(axes, trees, titles):
    pos = nx.spring_layout(tree, seed=42)
    nx.draw(tree, pos, ax=ax, with_labels=True, node_color='skyblue',
            node_size=500, edge_color='gray', font_weight='bold')
    ax.set_title(title)
 
plt.suptitle('Non-isomorphic Trees')
plt.tight_layout()
plt.show()
 
```

```python
import scipy
```

```python
def run_test(lb, pact, treelet_size):
    stat, p = scipy.stats.mannwhitneyu(lb, pact, alternative='less')
    print(f"{treelet_size}-treelet:")
    print(f"  U-statistic: {stat:.4f}")
    print(f"  p-value:     {p:.6f}")
    print(f"  Significant: {'Yes' if p < 0.05 else 'No'}")
    print()

```

```python
run_test(k7_regular["Lanzinger-Barcelo"], k7_regular["PACT"], 7)
run_test(k8_regular["Lanzinger-Barcelo"], k8_regular["PACT"], 8)
run_test(k9_regular["Lanzinger-Barcelo"], k9_regular["PACT"], 9)
```
